```python
import pandas as pd
```

```python
#Crear una variable que contenga el dataframe y mostrar las primeras clumnas de df
data_df = pd.read_csv("superstore_data.csv", delimiter=",")
data_df.head(5)
```

```python
#Mostrar tipo de dato de cada columna
data_df.dtypes
```

```python
data_df[data_df["Postal Code"].isna()].head(5)
```

```python
data_df_cp = data_df.copy()
data_df_cp["Postal Code"]= data_df_cp["Postal Code"].fillna(0)
data_df_cp[data_df["Postal Code"].isnull()].head(10)
```

```python
#Casteo al tipo de dato correcto en cada columna

data_df_cp["Postal Code"] = data_df_cp["Postal Code"].astype("float").astype("int").astype("string")
data_df_cp.dtypes
```

```python
#Casteo al tipo de dato correcto en cada columna
data_df_order_date = data_df_cp.copy()
data_df_order_date["Order Date"] = pd.to_datetime(
    data_df_order_date["Order Date"] , format="%d/%m/%Y"
)
data_df_order_date.dtypes
```

```python
data_df_ship_date = data_df_order_date.copy()
data_df_ship_date["Ship Date"] = pd.to_datetime(
    data_df_ship_date["Ship Date"] , format="%d/%m/%Y"
)
data_df_ship_date.dtypes
```

```python
data_df_ship_date.head()
```

```python
data_df_ship_date.shape
```

```python
data_df_names = data_df_ship_date.copy()
data_df_names["name"] = data_df_names["Customer Name"].str.split().str[0]
data_df_names["last_name"] = data_df_names["Customer Name"].str.split().str[1:].str.join(" ")
data_df_names.drop(columns=["Customer Name"], inplace=True)
data_df_names.head(5)
```
